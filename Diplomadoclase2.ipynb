```python
### Definir función
def interes_simple(capital, interes, tiempo):
  monto = capital* (1 + interes*tiempo)
  return(monto)
```

```python
interes_simple(capital=1000,interes=0.1,tiempo=12)
```

```python
interes_simple(1000,0.1,12)
```

```python

def interes_simple2(capital, interes, tiempo = 10):
  monto = capital* (1 + interes*tiempo)
  return(monto)



interes_simple2(capital = 1000, interes = 0.1)
```

```python

def interes_compuesto(capital, interes, tiempo):
  monto = capital* (1 + interes)**tiempo
  return(monto)



interes_compuesto(capital = 1000, interes = 0.1,tiempo=12)
```

```python

def anualidad_ordinaria(pago, interes, tiempo):
  monto = pago* (((1 + interes)**tiempo - 1))/interes
  return(monto)
```

```python
anualidad_ordinaria(pago=100,interes=0.01,tiempo=36)
```

```python
def cuota_mensual(capital, interes, cuotas):
  cuota=capital* (interes/(1-(1 + interes)**(-cuotas) ))
  return(cuota)
```

```python
cuota_mensual(capital=1000,interes=0.1,cuotas=12)
```

```python
mejor_truck = (resenas
    .groupby('truck_id')['calificacion']
    .mean()
    .reset_index()
    .merge(food_trucks[['truck_id','nombre_truck','tipo_cocina']], on='truck_id')
    .sort_values('calificacion', ascending=False)
)
```

```python
gasto = (pd.merge(pedidos, clientes[['cliente_id','nombre','apellido']], on='cliente_id')
           .groupby(['cliente_id','nombre','apellido'])['total_cop']
           .sum()
           .reset_index()
           .rename(columns={'total_cop':'gasto_total'}))

calificacion = (resenas.groupby('cliente_id')['calificacion']
                       .mean()
                       .reset_index()
                       .rename(columns={'calificacion':'calif_promedio'}))

pd.merge(gasto, calificacion, on='cliente_id').sort_values('gasto_total', ascending=False)
```

```python
compradores = pedidos['cliente_id'].value_counts().reset_index()
compradores.columns = ['cliente_id','num_pedidos']

resenas_ids = resenas[['cliente_id']].drop_duplicates().assign(tiene_resena=True)

(pd.merge(compradores, resenas_ids, on='cliente_id', how='left')
   .query('tiene_resena.isna()')
   .merge(clientes[['cliente_id','nombre','apellido']], on='cliente_id')
   .sort_values('num_pedidos', ascending=False))
```

```python
(pd.merge(resenas, food_trucks[['truck_id','nombre_truck']], on='truck_id')
   .assign(dia=lambda x: pd.to_datetime(x['fecha']).dt.day_name())
   .groupby(['nombre_truck','dia'])['calificacion']
   .mean()
   .reset_index()
   .sort_values(['nombre_truck','calificacion'], ascending=[True,False]))
```

```python
## Datos pokemon, correlación

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Seleccionar solo las columnas numéricas (excluyendo '#' y 'Legendary' como bool)
numeric_cols = ['Total', 'HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed', 'Generation']
correlation_matrix = df[numeric_cols].corr()

# Mostrar la matriz de correlación
print("Matriz de Correlación:")
print(correlation_matrix.round(3))
print()

# --- Gráfico 1: Heatmap de correlación ---
fig, ax = plt.subplots(figsize=(10, 8))
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool), k=1)
sns.heatmap(
    correlation_matrix,
    mask=mask,
    annot=True,
    fmt='.2f',
    cmap='RdBu_r',
    center=0,
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5,
    cbar_kws={'shrink': 0.8, 'label': 'Correlación'},
    ax=ax
)
ax.set_title('Correlación entre estadísticas Pokémon', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# --- Gráfico 2: Pairplot de las stats principales ---
stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
g = sns.pairplot(df[stats + ['Legendary']], hue='Legendary', diag_kind='kde',
                 plot_kws={'alpha': 0.5, 's': 15}, palette={True: '#e74c3c', False: '#3498db'})
g.figure.suptitle('Distribución y correlación por pares (Legendarios vs No Legendarios)', y=1.02,
                   fontsize=14, fontweight='bold')
plt.show()

# --- Gráfico 3: Top correlaciones ---
upper = correlation_matrix.where(mask.T)
pairs = upper.stack().reset_index()
pairs.columns = ['Var 1', 'Var 2', 'Correlación']
pairs['abs'] = pairs['Correlación'].abs()
top_pairs = pairs.sort_values('abs', ascending=False).head(10).drop(columns='abs')

fig, ax = plt.subplots(figsize=(10, 5))
colors = ['#e74c3c' if v > 0 else '#3498db' for v in top_pairs['Correlación']]
labels = top_pairs['Var 1'] + ' vs ' + top_pairs['Var 2']
ax.barh(labels, top_pairs['Correlación'], color=colors)
ax.set_xlabel('Coeficiente de correlación')
ax.set_title('Top 10 pares con mayor correlación (absoluta)', fontsize=14, fontweight='bold')
ax.axvline(x=0, color='gray', linewidth=0.5)
plt.tight_layout()
plt.show()
```
